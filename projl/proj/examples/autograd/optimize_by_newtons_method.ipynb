# 使用牛顿法寻找极小值

对 $f(x)=x^4-2x^2$ 计算一阶与二阶导数，按 $x ← x-f'(x)/f''(x)$ 更新。每轮输出更新前的参数。

选择安装了 `mydezero` 的 Python 3 内核，然后从上到下运行各单元格。

## 1. 导入依赖

```python
import numpy as np
from mydezero import Variable
```

## 2. 定义目标函数

```python
def f(x):
    y = x ** 4 - 2 * x ** 2
    return y
```

## 3. 设置初值与迭代次数

```python
x = Variable(np.array(2.0))
iters = 10
```

## 4. 使用二阶导数更新参数

```python
for i in range(iters):
    print(i, x)
    y = f(x)
    x.cleargrad()
    y.backward(create_graph=True)

    gx = x.grad
    x.cleargrad()
    gx.backward()
    gx2 = x.grad

    x.data -= gx.data / gx2.data
```
