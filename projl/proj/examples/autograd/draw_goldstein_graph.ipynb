# Goldstein Computation Graph

Evaluate the two-input Goldstein function and its gradients, then plot its graph. Requires Graphviz (`dot`); saves to `draft/goldstein.png`.

Select a Python 3 kernel with `mydezero` installed and run cells in order.

## 1. Import Dependencies

```python
import numpy as np
from mydezero import Variable
from mydezero.functions import goldstein
from mydezero.utils import plot_dot_graph
```

## 2. Evaluate, Backpropagate, and Plot

```python
x = Variable(np.array(1.0))
y = Variable(np.array(1.0))
z = goldstein(x, y)
z.backward()

x.name='x'
y.name = 'y'
z.name = 'z'

to_file = 'goldstein.png'
plot_dot_graph(z, verbose=False, to_file=to_file)
```
