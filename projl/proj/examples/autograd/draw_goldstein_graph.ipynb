# Goldstein 函数的计算图

计算二元 Goldstein 函数及其梯度，并用 Graphviz 展示计算图。需要系统已安装 Graphviz（`dot` 命令）；图片保存在项目的 `draft/goldstein.png`。

选择安装了 `mydezero` 的 Python 3 内核，然后从上到下运行各单元格。

## 1. 导入依赖

```python
import numpy as np
from mydezero import Variable
from mydezero.functions import goldstein
from mydezero.utils import plot_dot_graph
```

## 2. 计算函数、反向传播并绘图

```python
x = Variable(np.array(1.0))
y = Variable(np.array(1.0))
z = goldstein(x, y)
z.backward()

x.name='x'
y.name = 'y'
z.name = 'z'

to_file = 'goldstein.png'
plot_dot_graph(z, verbose=False, to_file=to_file)
```
