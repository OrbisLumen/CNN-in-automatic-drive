```python
import numpy as np
from mydezero import Variable
```

```python
def f(x):
    y = x ** 4 - 2 * x ** 2
    return y
```

```python
x = Variable(np.array(2.0))
iters = 10
```

```python
for i in range(iters):
    print(i, x)
    y = f(x)
    x.cleargrad()
    y.backward(create_graph=True)

    gx = x.grad
    x.cleargrad()
    gx.backward()
    gx2 = x.grad

    x.data -= gx.data/ gx2.data
```
